# Distributed Training

This experiment runs the same online FedAvg trajectory with ZipNN and NeuralZip communication measurements. NeuralZip builds its setup once from the initial model and reuses it for all 200 rounds.

```python
import os
from pathlib import Path

from IPython.display import display

from neuralzip.experiment_scenarios import run_distributed_experiment_suite


ARCHITECTURES = ("resnet18", "resnet34", "vgg16_bn", "vit_b_16")
OUTPUT_DIR = Path(os.environ.get("NEURALZIP_RUN_ROOT", "outputs/distributed/cifar100_200x10"))
DATA_DIR = Path(os.environ.get("NEURALZIP_CIFAR100_DIR", "outputs/datasets/cifar100_hf"))
protocol_rounds, protocol_summary = run_distributed_experiment_suite(
    OUTPUT_DIR,
    architectures=ARCHITECTURES,
    data_dir=DATA_DIR,
    clients=10,
    rounds=200,
    local_batches=1,
    batch_size=64,
)
display(protocol_summary)
display(protocol_rounds)
```
