# Checkpoints

This experiment trains a small ResNet-18 for a few epochs on CIFAR-shaped synthetic data, stores every checkpoint, and then compresses the saved checkpoints. Training is meant to run on the local GPU machine, while the compression cell can be executed on `godel` over the generated checkpoint files.

```python
from pathlib import Path

from IPython.display import display

from neuralzip.experiment_scenarios import train_resnet18_cifar_checkpoints


OUTPUT_DIR = Path("outputs/checkpoints/resnet18_cifar10")
training_metrics = train_resnet18_cifar_checkpoints(
    OUTPUT_DIR,
    epochs=3,
    train_subset=5000,
    test_subset=1000,
    batch_size=128,
    dataset="fake",
)
display(training_metrics)
```

```python
from pathlib import Path

from IPython.display import display

from neuralzip.experiment_scenarios import DEFAULT_METHODS, compress_checkpoint_directory


OUTPUT_DIR = Path("outputs/checkpoints/resnet18_cifar10")
checkpoint_results = compress_checkpoint_directory(
    OUTPUT_DIR / "checkpoints",
    OUTPUT_DIR / "compression",
    methods=DEFAULT_METHODS,
)
display(checkpoint_results)
```
