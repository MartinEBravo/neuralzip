# Vision and diffusion models

This notebook runs the non-language architecture suite with the same bf16 exponent-stream compression scope used by the large-checkpoint experiments.

```python
from pathlib import Path
import sys

cwd = Path.cwd().resolve()
if (cwd / "config.py").exists():
    root = cwd.parent
    output_dir = Path("outputs/vision_diffusion")
else:
    root = cwd
    output_dir = Path("examples/outputs/vision_diffusion")

for path in (root, root / "examples"):
    if str(path) not in sys.path:
        sys.path.insert(0, str(path))

from config import VISION_DIFFUSION_MODEL_SUITE
from neuralzip.experiment_scenarios import DEFAULT_METHODS, run_vision_diffusion_experiment

vision_diffusion_results = run_vision_diffusion_experiment(
    output_dir,
    model_suite=VISION_DIFFUSION_MODEL_SUITE,
    methods=DEFAULT_METHODS,
    local_files_only=False,
    resume=True,
)
vision_diffusion_results
```

```python
import pandas as pd

ok = vision_diffusion_results[vision_diffusion_results["status"].eq("ok")].copy()
time_table = ok[ok["compressor"].isin(["ZipNN local", "NeuralZip"])].pivot(
    index="model_name",
    columns="compressor",
    values="encode_seconds",
)
ratio_table = ok[ok["compressor"].isin(["ZipNN local", "NeuralZip"])].pivot(
    index="model_name",
    columns="compressor",
    values="compressed_pct",
)
display(time_table.round(2))
display(ratio_table.round(2))
```
