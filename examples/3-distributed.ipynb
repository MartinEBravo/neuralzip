# Federated Learning

This experiment simulates FedAvg with compressed communication. FedZipNN compresses and decompresses every server-to-client and client-to-server message with local ZipNN codebooks. FedNeuralZip builds shared tensor codebooks once, sends them to the clients, and reuses them for all compressed messages across rounds.

```python
from pathlib import Path

from IPython.display import display

from neuralzip.experiment_scenarios import run_federated_codec_protocol_online


OUTPUT_DIR = Path("outputs/distributed/fedavg_resnet18_cifar10_200x10")
protocol_rounds, protocol_summary = run_federated_codec_protocol_online(
    OUTPUT_DIR,
    clients=10,
    rounds=200,
    local_batches=1,
    batch_size=64,
    dataset="hf-cifar10",
    data_dir=Path("outputs/datasets/cifar10_hf"),
)
display(protocol_summary)
display(protocol_rounds)
```
