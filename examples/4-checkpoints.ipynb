# Checkpoints

This experiment trains four CIFAR-100 architectures for 200 epochs, stores epochs 0 through 200, and compresses every checkpoint independently with NeuralZip and the ZipNN baseline.

```python
import os
from pathlib import Path

from IPython.display import display

from neuralzip.experiment_scenarios import run_checkpoint_experiment_suite


ARCHITECTURES = ("resnet18", "resnet34", "vgg16_bn", "vit_b_16")
OUTPUT_DIR = Path(os.environ.get("NEURALZIP_RUN_ROOT", "outputs/checkpoints/cifar100_200_epochs"))
DATA_DIR = Path(os.environ.get("NEURALZIP_CIFAR100_DIR", "outputs/datasets/cifar100_hf"))
training_metrics, checkpoint_results, checkpoint_summary = run_checkpoint_experiment_suite(
    OUTPUT_DIR,
    architectures=ARCHITECTURES,
    data_dir=DATA_DIR,
    epochs=200,
    batch_size=128,
)
display(training_metrics)
display(checkpoint_summary)
display(checkpoint_results)
```
